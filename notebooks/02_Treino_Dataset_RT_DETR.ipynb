```python
from ultralytics import YOLO

# no final do treinamento, YOLO salva dois arquivos:
#    - last.pt (o último momento do treino)
#    - best.pt (o momento em que o modelo teve a melhor precisão)
model = YOLO('/content/runs/detect/train/weights/best.pt')

# converte o cérebro do modelo para o formato TFLite (TensorFlow Lite), que é a linguagem universal de IA para dispositivos móveis
# usando int8=True ajuda a deixar o modelo ainda mais leve para celular
model.export(format='tflite', int8=True)
```
