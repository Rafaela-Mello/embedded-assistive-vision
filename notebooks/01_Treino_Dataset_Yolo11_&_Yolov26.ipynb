# **Código Completo**

```python
!pip install roboflow

# classe principal que gerencia a conexão e as requisições de dados
from roboflow import Roboflow
rf = Roboflow(api_key="SUA_API_KEY")
project = rf.workspace("seu-workspace").project("seu-projeto")
version = project.version(1)
dataset = version.download("yolov11") # já garante que me dê o arquivo no formato exato que o YOLO11 espera: .txt

# contém o código necessário para treinar, validar e usar os modelos YOLO
!pip install ultralytics

print("O caminho do meu dataset é:", dataset.location)

from ultralytics import YOLO

# modelo "pré-treinado": já sabe reconhecer formas básicas (bordas, cores, curvas) porque já viu milhões de fotos antes
model = YOLO('yolo11n.pt') # carrega o modelo YOLO11 nano (mais leve e rápido)

# inicia o treino
# data: aponta para o arquivo .yaml que o Roboflow baixou automaticamente (onde estão as fotos e quais são as etiquetas)
# epochs: o modelo vai passar 50 vezes por todo o seu conjunto de fotos para aprender
# imgsz: redimensiona todas as fotos para 640x640 pixels (o padrão de equilíbrio entre precisão e velocidade)
results = model.train(data=f"{dataset.location}/data.yaml", epochs=50, imgsz=640)
```

```python
# Código Python para rodar a Webcam: Crie um arquivo chamado testar_webcam.py na mesma pasta onde está o seu best.pt:

import cv2
from ultralytics import YOLO

# 1. Carrega o modelo treinado
model = YOLO('best.pt')

# 2. Abre a conexão com a webcam (0 é o índice da webcam padrão)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Erro ao abrir a webcam")
    exit()

print("Pressione 'q' para sair.")

while True:
    # Captura frame por frame
    ret, frame = cap.read()
    if not ret:
        break

    # 3. Executa a detecção do YOLO11 no frame atual
    # conf=0.5 ignora detecções com menos de 50% de certeza
    results = model(frame, conf=0.5)

    # 4. Desenha os resultados no frame (caixas, nomes e porcentagem)
    annotated_frame = results[0].plot()

    # 5. Exibe o vídeo em uma janela
    cv2.imshow('YOLO11 Real-Time Detection', annotated_frame)

    # Fecha se pressionar a tecla 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Limpeza
cap.release()
cv2.destroyAllWindows()
```
